# Data understanding
Synthetic UPI-like transaction data generated for educational purposes.

```python
import pandas as pd
df = pd.read_csv('../data/processed/transactions.csv', parse_dates=['timestamp'])
display(df.head(), df.dtypes, df.isna().sum())
```

```python
summary = {'shape': df.shape, 'users': df.user_id.nunique(), 'receivers': df.receiver_id.nunique(), 'date_range': (df.timestamp.min(), df.timestamp.max()), 'fraud_count': int(df.is_fraud.sum()), 'fraud_pct': df.is_fraud.mean()*100}
summary
```
