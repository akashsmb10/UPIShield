# Modeling
Isolation Forest is unsupervised: it learns which feature combinations are easy to isolate without seeing labels. Logistic Regression and Random Forest are supervised and learn from fraud labels.

The chronological 70/15/15 split prevents future leakage. Preprocessing fits on training; validation selects the model and thresholds; test is final-only. Under class imbalance, precision, recall, F1 and PR-AUC are more informative than accuracy. The hybrid combines supervised probability (50%), anomaly score (30%), and explainable rules (20%).
